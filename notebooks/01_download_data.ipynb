# Download Data

Download SST-2 and CSD Negation datasets.


```python
import sys
import os
from pathlib import Path

# Ensure we're in the right directory
if not os.path.exists('src'):
    if os.path.exists('Negation-Origin-Tracing'):
        os.chdir('Negation-Origin-Tracing')
    else:
        print("⚠ Warning: Could not find project directory")
        print(f"  Current: {os.getcwd()}")
        print(f"  Run 00_setup_colab.ipynb first!")

sys.path.insert(0, os.getcwd())
print(f"✓ Working directory: {os.getcwd()}")
```

```python
# Create data directories
os.makedirs('data/raw/train', exist_ok=True)
os.makedirs('data/raw/validation', exist_ok=True)
os.makedirs('data/raw/test', exist_ok=True)
print("✓ Data directories created")
```

```python
# Download SST-2 dataset
from datasets import load_dataset

print("Downloading SST-2 dataset...")
train = load_dataset("stanfordnlp/sst2", split="train")
validation = load_dataset("stanfordnlp/sst2", split="validation")
test = load_dataset("stanfordnlp/sst2", split="test")

train.to_parquet("data/raw/train/sst.parquet")
validation.to_parquet("data/raw/validation/sst.parquet")
test.to_parquet("data/raw/test/sst.parquet")

print(f"✓ SST-2 downloaded: {len(train)} train, {len(validation)} val, {len(test)} test")
```

```python
# Download CSD Negation dataset
try:
    print("Downloading CSD Negation dataset...")
    train_neg = load_dataset("ceval/contrastive-sentiment-negation", split="train")
    validation_neg = load_dataset("ceval/contrastive-sentiment-negation", split="validation")
    test_neg = load_dataset("ceval/contrastive-sentiment-negation", split="test")
    
    train_neg.to_parquet("data/raw/train/negation.parquet")
    validation_neg.to_parquet("data/raw/validation/negation.parquet")
    test_neg.to_parquet("data/raw/test/negation.parquet")
    
    print(f"✓ CSD Negation downloaded: {len(train_neg)} train, {len(validation_neg)} val, {len(test_neg)} test")
except Exception as e:
    print(f"⚠ Could not download CSD Negation dataset: {e}")
    print("  You can continue with SST-2 only")
```

```python
# Verify data files
import pandas as pd

print("\nVerifying data files...")
for split in ['train', 'validation', 'test']:
    sst_path = f"data/raw/{split}/sst.parquet"
    if os.path.exists(sst_path):
        df = pd.read_parquet(sst_path)
        print(f"✓ {split}/sst.parquet: {len(df)} samples")
        print(f"  Columns: {df.columns.tolist()}")
        if len(df) > 0:
            print(f"  Sample: {df.iloc[0]['sentence'][:50]}...")
```
