```python
# Configuration
results_path = 'experiments/layer_search/results_summary.json'
output_dir = 'experiments/layer_search/visualizations'

print(f"Results path: {results_path}")
print(f"Output directory: {output_dir}")
```

```python
# Display plots inline
from IPython.display import Image, display

if os.path.exists(output_dir):
    plot_files = [
        'layer_performance_test_auroc.png',
        'pooling_comparison_test_auroc.png',
        'heatmap_test_auroc.png',
        'best_layers_test_auroc.png',
    ]
    
    for plot_file in plot_files:
        plot_path = os.path.join(output_dir, plot_file)
        if os.path.exists(plot_path):
            print(f"\n{plot_file}:")
            display(Image(plot_path))
```
